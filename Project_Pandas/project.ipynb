```python
#Coverting Billing Amout float into int
# Converting Billing Amount float into int
df['Billing Amount'] = pd.to_numeric(
    df['Billing Amount'], errors='coerce'
).round().astype('Int64')

df.dtypes
df.dtypes
```

```python
#duplicate Values
df[df.duplicated()]
df=df.drop_duplicates()


#delete all negative billing values
df=df[df['Billing Amount']>=0]
print(df)
```

```python
#Fixing Inconsistent Date
df['Name']= df['Name'].str.title()
df.head()
```

```python
#Date 
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

df['Length of Stay'] = (
    df['Discharge Date'] - df['Date of Admission']
).dt.days

df.head()
```

```python
#Writing a funcion get_age_group(age)

def get_age_group(age):
    if age <= 12:
        return "Child"
    elif age <= 19:
        return "Teen"
    elif age <= 59:
        return "Adult"
    else:
        return "Senior"

df['Age Group']=df['Age'].apply(get_age_group)
df.head()
```

```python
#Sorting on Basis of Billing Amount in descending Order

df=df.sort_values(by='Billing Amount', ascending=[False])
df.head()
```

```python
#GroupBy and Aggregation by Medical Condition the Average of Billing Amount and the average length of Stay
df.groupby('Medical Condition')['Billing Amount'].mean()
```

```python
df.groupby('Medical Condition')['Length of Stay'].mean()
```

```python
conditional_summary = df.groupby('Medical Condition').agg(
    bill_avg = ('Billing Amount','mean'),
    days_avg = ('Length of Stay','mean')
)

print(conditional_summary)
```
